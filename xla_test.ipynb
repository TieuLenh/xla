```python
# Bài 1 — Histogram, Brightness, Contrast, Histogram Equalization
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Doc anh grayscale
image = cv2.imread("test.jpg", cv2.IMREAD_GRAYSCALE)
if image is None:
    raise ValueError("Khong doc duoc anh")

def histogram(img):
    h = [0] * 256
    for x in img:
        for y in x:
            h[y] += 1
    return h

def show_hist(img):
    hist_arr = histogram(img)
    x = np.arange(0, 256)
    plt.bar(x, hist_arr)
    plt.show()

def adjust_brightness(img, brightness):
    new_img = np.copy(img).astype(int) + brightness
    for i in range(len(new_img)):
        for j in range(len(new_img[0])):
            if new_img[i][j] > 255:
                new_img[i][j] = 255
            if new_img[i][j] < 0:
                new_img[i][j] = 0
    return new_img.astype(np.uint8)

def incresse_brightness(img):
    brightness = int(input('Nhap do sang muon tang: '))
    return adjust_brightness(img, brightness= brightness)

def decrease_brightness(img):
    brightness = int(input('Nhap do sang muon giam: '))
    return adjust_brightness(img, brightness= -brightness)

# main
show_hist(img= image)

ib_img = incresse_brightness(img= image)
db_img = decrease_brightness(img= image)

cv2.imshow("Increase Brightness", ib_img)
cv2.imshow('Decrease bightness', db_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import cv2
import numpy as np

img = cv2.imread( "test.jpg",cv2.IMREAD_GRAYSCALE)

# Kernel trung bình
kernel = np.ones((3,3), np.float32) / 9
# 1. Convolution
conv = cv2.filter2D(img, -1, kernel)
# 2. Median
median = cv2.medianBlur( img, 3)
# 3. Mean
mean = cv2.blur( img, (3,3))

cv2.imshow("Original", img)
cv2.imshow("Convolution", conv)
cv2.imshow("Median", median)
cv2.imshow("Mean", mean)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Bài 5 — KMeans Segmentation
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Doc anh mau
image = cv2.imread("test.jpg")
if image is None:
    raise ValueError("Khong doc duoc anh")

# Chuyen BGR -> RGB
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# Dua anh ve dang vector
# Moi pixel = 1 diem du lieu
pixel_values = image_rgb.reshape((-1, 3))
pixel_values = np.float32(pixel_values)

# Dieu kien dung
criteria = (cv2.TERM_CRITERIA_EPS + cv2.TERM_CRITERIA_MAX_ITER, 100, 0.2)

# KMeans
k = 3
_, labels, centers = cv2.kmeans( pixel_values, k, None, criteria, 10, cv2.KMEANS_RANDOM_CENTERS)

# Tao anh moi tu center
centers = np.uint8(centers)
labels = labels.flatten()

segmented = centers[labels]
segmented = segmented.reshape(image_rgb.shape)

# Hien thi
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(image_rgb)
plt.title("Original")
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(segmented)
plt.title("KMeans")
plt.axis('off')

plt.tight_layout()
plt.show()
```
